# The Transformation Logic

```python
query = """
SELECT 
  SALES.order_number,
  SALES.product_number,
  PRODUCTS.product_key,
  CUSTOMERS.customer_key,
  SALES.customer_id,
  SALES.order_date,
  SALES.ship_date,
  SALES.due_date,
  SALES.sales,
  SALES.quantity,
  SALES.price
FROM workspace.silver.crm_sales AS SALES
LEFT JOIN workspace.silver.crm_products AS PRODUCTS
ON SALES.product_number = PRODUCTS.product_key
LEFT JOIN workspace.silver.crm_customers AS CUSTOMERS
ON SALES.customer_id = CUSTOMERS.customer_id
"""

df = spark.sql(query)
df.display()
```

# Save Gold Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```

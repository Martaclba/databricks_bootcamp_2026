# The Transformation Logic

```python
query = """
SELECT 
    ROW_NUMBER() OVER (ORDER BY PRODUCTS.product_start_date, PRODUCTS.product_number) AS product_rank,
    PRODUCTS.product_id,
    PRODUCTS.product_key,
    PRODUCTS.category_id,
    PRODUCTS.product_number,
    PRODUCTS.product_cost,
    PRODUCTS.product_line,
    PRODUCTS.product_start_date,
    PRODUCTS.product_end_date,
    PRODUCT_CATEGORY.category_name,
    PRODUCT_CATEGORY.subcategory_name,
    PRODUCT_CATEGORY.maintenance_flag
FROM workspace.silver.crm_products AS PRODUCTS
LEFT JOIN workspace.silver.erp_product_category AS PRODUCT_CATEGORY
ON PRODUCTS.category_id = PRODUCT_CATEGORY.category_id
"""

df = spark.sql(query)
df.display()
```

# Save Gold Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_products")
```

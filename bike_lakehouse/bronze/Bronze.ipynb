# Read From CSV and Write Silver Table

```python
for item in INGESTION_CONFIG:
    print(f"Ingesting {item['table']} from source: {item['source']}]")
    df = (
        spark.read
             .option("header", "True")
             .option("inferSchema", "True")
             .csv(item["path"])
    )

    df.write.mode("overwrite").format("delta").saveAsTable(item["table"])
```
